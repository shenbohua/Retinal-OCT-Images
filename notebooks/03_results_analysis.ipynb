# 03 Results Analysis

Analyse baseline matrix outputs and generate report-ready tables/plots.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

PROJECT_ROOT = Path.cwd().parent
TABLES_DIR = PROJECT_ROOT / 'outputs' / 'tables'
FIG_DIR = PROJECT_ROOT / 'outputs' / 'figures'
FIG_DIR.mkdir(parents=True, exist_ok=True)

results_path = TABLES_DIR / 'traditional_results.csv'
if not results_path.exists():
    raise FileNotFoundError('Run `python main.py train-matrix` first.')

results_df = pd.read_csv(results_path)
results_df.sort_values('macro_f1', ascending=False).head(10)
```

```python
pivot = results_df.pivot_table(index='feature', columns='classifier', values='macro_f1', aggfunc='max')
pivot
```

```python
plt.figure(figsize=(7, 4))
sns.heatmap(pivot, annot=True, fmt='.3f', cmap='YlGnBu')
plt.title('Macro-F1 Heatmap (Traditional Matrix)')
plt.tight_layout()
plt.savefig(FIG_DIR / 'traditional_macro_f1_heatmap.png', dpi=180)
plt.show()
```

```python
report_cols = [
    'feature', 'classifier', 'accuracy', 'macro_f1',
    'macro_precision', 'macro_recall', 'train_time_sec',
    'inference_time_ms_per_image', 'notes'
]
report_table = results_df[report_cols].sort_values('macro_f1', ascending=False)
report_table.to_csv(TABLES_DIR / 'traditional_results_for_report.csv', index=False)
report_table.head(10)
```
